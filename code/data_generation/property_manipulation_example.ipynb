## Use Thin Lens Camera

```python
id_model = 30
id_hairModel = 62
id_lesion = 6
sel_lesionMat = 'HbO2x1.0Epix0.25'
sel_lightName = 'diffuse'
sel_hair_albedo = [0.84, 0.6328, 0.44]
id_fracBlood = 0.002
id_mel = 0.25
id_timePoint = 35
res = 32
id_origin_y = 13.0
lesion_directory = '../../data/supporting_data/materials/lesions_release/ver2/'
skin_layers_directory = '../../data/supporting_data/materials/outputModels/'

scene_ref = util.render_image(id_model, id_hairModel, id_lesion,
             sel_lesionMat, sel_lightName, sel_hair_albedo, 
             id_fracBlood, id_mel,
             id_timePoint, id_origin_y,
             IMAGE=True,
             lesion_directory=lesion_directory, skin_layers_directory=skin_layers_directory,
             pre_processed_lesion=True,  verbose=True)

def get_sensor_thin_lens(id_origin_y):
    # Creating a single sensor from top
    cam_top = mi.load_dict({
        'type': 'thinlens', 
        'focal_length': '24mm', 
        'srf': {
              'type': 'uniform',
              'value': 1.0 
        },
        'to_world': mi.scalar_spectral.Transform4f.look_at(
            target=[0, 0, 0],
            origin=[0, id_origin_y, 0], #[0, 15, 0],
            up=[0, 0, 1]),  
        'focus_distance': 10.0,
        'aperture_radius': 1.85,
        'film': {
            'type': 'hdrfilm',
            'width': 1024, 'height': 1024,
        }
    })
    return cam_top


cam_top_thinLens = get_sensor_thin_lens(id_origin_y)

ref_image = mi.render(scene_ref, sensor=cam_top_thinLens, spp=res)
ref_image_bitmap = mi.util.convert_to_bitmap(ref_image)
plt.imshow(ref_image_bitmap)
plt.axis('off')
plt.show()
```
